```python
import numpy as np
import scipy.stats as stats
from scipy.stats import t, normb
import math
```

```python
# Данные задачи А
data_A = np.array([6.9, 11.2, 13.8, 11.6, 10.7, 13.5, 9.9, 8.7, 8.2, 10.8, 8.6, 7.7])
n_A = len(data_A)
x_bar_A = np.mean(data_A)
s_A = np.std(data_A, ddof=1)

print(f"n = {n_A}, среднее = {x_bar_A:.4f}, std = {s_A:.4f}")
```

```python
e_A = 0.5
t_stat_A = e_A * np.sqrt(n_A) / s_A
conf_prob_A = 2 * t.cdf(t_stat_A, df=n_A-1) - 1
print(f"t = {t_stat_A:.4f}")
print(f"Доверительная вероятность = {conf_prob_A:.4f} ({conf_prob_A*100:.2f}%)")
```

```python
e_B = 0.25
conf_level = 0.95
alpha = 1 - conf_level

z = norm.ppf(1 - alpha/2)
n_est = (z * s_A / e_B) ** 2
n_cur = max(2, int(np.ceil(n_est)))

while True:
    t_crit = t.ppf(1 - alpha/2, df=n_cur-1)
    n_new = (t_crit * s_A / e_B) ** 2
    n_new = max(2, int(np.ceil(n_new)))
    if n_new == n_cur:
        break
    n_cur = n_new

print(f"Необходимое число измерений: n = {n_cur}")
```
